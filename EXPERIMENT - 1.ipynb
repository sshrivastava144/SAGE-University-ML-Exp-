Experiment: Predicting House Prices Based on Size
We'll use a small dataset where house prices depend on the size of the house. The
model will learn from this data and predict the price for a given house size.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Sample Data: House Size (sq ft) vs. House Price (in $1000s)
X = np.array([500, 700, 800, 1000, 1200, 1500, 1800, 2000, 2200, 2500]).reshape(-1, 1)
y = np.array([100, 150, 180, 200, 240, 270, 300, 330, 360, 400])

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Predict price for a new house size (e.g., 1600 sq ft)
new_size = np.array([[1600]])
predicted_price = model.predict(new_size)

# Print the equation of the line
print(f"Equation of the line: y = {model.coef_[0]:.2f}x + {model.intercept_:.2f}")

# Print the predicted price
print(f"Predicted price for a 1600 sq ft house: ${predicted_price[0]:.2f}K")

# Visualization
plt.scatter(X, y, color='blue', label="Actual Data")
plt.plot(X, model.predict(X), color='red', linestyle='--', label="Regression Line")
plt.scatter(new_size, predicted_price, color='green', marker='o', s=100, label="Prediction")
plt.xlabel("House Size (sq ft)")
plt.ylabel("House Price ($1000s)")
plt.title("Simple Linear Regression: House Price Prediction")
plt.legend()
plt.show()
```
